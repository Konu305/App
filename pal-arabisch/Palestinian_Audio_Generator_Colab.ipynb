## Ergebnis prüfen
Im Ordner `pal_audio` liegen für jeden Satz eine normale und eine langsame WAV-Datei sowie `manifest.json` und `REVIEW.tsv`.

Höre zunächst einige Beispiele an. Besonders wichtig sind Kaffee, Kiosk, Verstehen, Begrüßung und Wegfragen.

```python
import json
from pathlib import Path
from IPython.display import Audio, display

manifest = json.loads(Path('pal_audio/manifest.json').read_text())
sample_ids = list(manifest)[:5]
for phrase_id in sample_ids:
    print(phrase_id)
    display(Audio('pal_audio/' + manifest[phrase_id]['normal']))
```
